```python
import numpy as np 
import matplotlib.pyplot as plt 
import word2vec as WV
%matplotlib inline
```

```python
createdict = False
loaddict = True
if createdict:
    WV.build_word_vec_dict('../data/GoogleNews-vectors-negative300.bin')
    
if loaddict:
    wordvecdict = WV.load_word_vec_dict('../data/wordvec_dict.p')
```

```python
allkeys = wordvecdict[0].keys()
allnumbers = wordvecdict[0].values()

def Word2Vec(word):
    return wordvecdict[0][word]
   

def Words2Array(words):
    import string
    words = words.translate(None, string.punctuation)
    words = words.split()
    array = []
    for i in range(len(words)):
        if words[i] in allkeys:
            array.append(Word2Vec(words[i]))
    array = np.array(array)
    return np.transpose(array)

def Vec2Word(vec):
    i = closest_node(vec, allnumbers)
    return allkeys[i]

def closest_node(node, nodes):
    nodes = np.asarray(nodes)
    deltas = nodes - node
    dist_2 = np.einsum('ij,ij->i', deltas, deltas)
    return np.argmin(dist_2)

def getTopic(w, iii=1):
    arr = Words2Array(w)
    mean = np.mean(arr, axis=1)
    print "Shape of arr: ", arr.shape
    cov = np.cov(arr)

    from scipy.linalg import eigh, eig

    EV,Evec = eigh(cov)
    idx = EV.argsort()[::-1]   
    EV = EV[idx]
    Evec = Evec[:,idx]
    
    topic = []
    for i in range(iii):
        Primevec = Evec[:,i]*EV[i]**0.5 + mean
        topic.append(Vec2Word(Primevec))
        print i, topic[i]
    return topic
```

```python
sent2 = "There are only two ways to live your life. One is as though nothing is a miracle. The other is as though everything is a miracle."
print getTopic(sent2, 5)
```
